# Gradient Descent — Finding Local Minimum
Implements Gradient Descent to find the local minimum of the function $f(x) = x^2 + 4x + 5$ (analytical minimum at $x = -2, f(-2) = 1$).

### 1. Define Objective Function and Derivative

```python
def f(x):
    return x * x + 4 * x + 5

def derivative(x):
    return 2 * x + 4
```

### 2. Run Gradient Descent Optimization

```python
x = float(input("Initial x: "))
learning_rate = float(input("Learning rate (e.g. 0.1): "))
iterations = int(input("Number of iterations: "))

for i in range(iterations):
    x -= learning_rate * derivative(x)
    if i % max(1, iterations // 10) == 0:
        print(f"Iteration {i+1}: x={x:.6f}, f(x)={f(x):.6f}")

print(f"\nApproximate minimum: x={x:.6f}, f(x)={f(x):.6f}")
```
